# Process Census Data (2000, 2020)

```python
keep_cols_2000 = ["YEAR", "COUNTY", "CTY_SUB", "TOTAL", "WHITE", "BLACK", "NATIVE AMERICAN", "ASIAN", "PACIFIC ISLANDER", "OTHER0", "OTHER1",]
keep_cols_2020 = ["YEAR", "COUNTY", "COUSUB", "TOTAL", "WHITE", "BLACK", "NATIVE AMERICAN", "ASIAN", "PACIFIC ISLANDER", "OTHER0", "OTHER1",]
target_counties = ["Alameda", "Contra Costa", "Marin", "Napa", "San Mateo", "Santa Clara", "Solano", "Sonoma", "San Francisco"]
county_to_FIPS = {
    "Alameda" : "06001",
    "Contra Costa" : "06013",
    "Marin" : "06041",
    "Napa" : "06055",
    "San Mateo" : "06081",
    "Santa Clara" : "06085",
    "Solano" : "06095",
    "Sonoma" : "06097",
    "San Francisco" : "06075",
}

census_2000_df = pd.read_csv("demographic_raw/census/2000_census.csv")
label_to_race_2000 = {
    "GQO001":    "WHITE",
    "GQO002":    "BLACK",
    "GQO003":    "NATIVE AMERICAN",
    "GQO004":    "ASIAN",
    "GQO005":    "PACIFIC ISLANDER",
    "GQO006":    "OTHER0",
    "GQO007":    "OTHER1",
}

census_2000_df.rename(columns=label_to_race_2000, inplace=True)
census_2000_df["TOTAL"] = census_2000_df["WHITE"] + census_2000_df["BLACK"] + census_2000_df["ASIAN"] + census_2000_df["NATIVE AMERICAN"] + census_2000_df["PACIFIC ISLANDER"] + census_2000_df["OTHER0"] +census_2000_df["OTHER1"]
census_2000_df = census_2000_df[keep_cols_2000]
census_2000_df.rename(columns={"CTY_SUB" : "COUSUB"}, inplace = True)

census_2020_df = pd.read_csv("demographic_raw/census/2020_census.csv")
label_to_race_2020 = {
    "U7J001":    "TOTAL",
    "U7J002":    "WHITE",
    "U7J003":    "BLACK",
    "U7J004":    "NATIVE AMERICAN",
    "U7J005":    "ASIAN",
    "U7J006":    "PACIFIC ISLANDER",
    "U7J007":    "OTHER0",
    "U7J008":    "OTHER1",
}
census_2020_df.rename(columns=label_to_race_2020, inplace=True)
census_2020_df = census_2020_df[keep_cols_2020]
census_2020_df["COUNTY"] = census_2020_df["COUNTY"].str.removesuffix(" County")

census_combined = pd.concat([census_2000_df, census_2020_df], ignore_index=True)
census_combined["OTHER"] = census_combined["OTHER0"] + census_combined["OTHER1"] 
census_combined = census_combined.drop(["OTHER0", "OTHER1"], axis=1)
census_combined["COUSUB"] = census_combined["COUSUB"].str.removesuffix(" CCD")
census_combined = census_combined[census_combined["COUNTY"].isin(target_counties)]
census_combined["COUNTY_FIPS"] = census_combined["COUNTY"].map(county_to_FIPS)
census_combined = census_combined.iloc[:, [0, 1, 10, 2, 3, 4, 5, 6, 7, 8,9]]

print(census_combined.to_string())
census_combined.to_csv("demographic_processed/census_combined.csv", index = False)
```
